When we did a train/test split, the purpose was to evaluate our model on unseen data. However, once we are done with testing and want to actually use the model (for example, to predict the test molecules), we might want to retrain it on the entire dataset. Training on all the data could potentially improve performance, since the model sees more examples, but the downside is that we no longer have a test set to estimate its accuracy.

In this course, the performance estimate can also come from the public score on the test molecules (from Iteration 3), which serves as feedback on how well your model does on molecules it has never seen.
