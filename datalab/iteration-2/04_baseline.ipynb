### The metric

The Kaggle stability competition is scored with **macro-F1**: the F1 scores of `low`, `medium` and `high`, averaged with equal weight (see section 8 of [Modeling & Evaluation](../../self-study/007_modeling_and_evaluation.ipynb)). Accuracy is easy to read, but it can look good while a model ignores the smaller classes. In this notebook, report **accuracy and macro-F1** side by side (`f1_score(y_test, y_pred, average="macro")`), and use macro-F1 to make decisions. Always compare with a **majority-class baseline**, a "model" that always predicts the most frequent class: `DummyClassifier(strategy="most_frequent")` from `sklearn.dummy`.
