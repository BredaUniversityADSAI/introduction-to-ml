Now that we have parsed the text file, cleaned out empty values, and converted the columns to the correct data types, we can take a closer look at the dataset. While the data descriptor document specifies what each field represents, the meanings may not be immediately clear without a chemistry background. Below is a plain-language explanation of each column, prepared with the help of ChatGPT.

> **About the values in this file.** The molecules and the columns that describe the experiments (DOI, InChIKEY, construction, architecture, complement) are as published, and so is the fill factor. HOMO, LUMO and the electrochemical gap are synthetically generated, like all target values in this course (see the [Data Understanding](https://ai.buas.nl/open-courses/introduction-to-ml/datalab/iteration-1/02_data_understanding.html) page of Iteration 1). The optical gap, PCE, VOC and JSC are the published values, adjusted to the synthetic HOMO and LUMO of each molecule, so that the relations between the columns stay realistic. None of these numbers is the published measurement, so you will not find them in the papers.

1. **Digital Object Identifier (DOI)**

   * A permanent identifier of the research paper where the experiment was published.
   * Useful for tracing back to the original experimental setup and conditions.

2. **InChIKEY of molecule**

   * A hashed version of the InChI (International Chemical Identifier) string.
   * Provides a unique, compact ID for the molecule.
   * Unlike SMILES, it is not used for feature extraction. Normally it identifies a molecule and links it to chemical databases, but in this file it does not match the SMILES of its row in most cases: RDKit computes the same key from the SMILES for only 9 of the 350 rows. Identify the molecules by their canonical SMILES instead (see [Understanding SMILES](https://ai.buas.nl/open-courses/introduction-to-ml/datalab/iteration-2/01_understanding_smiles.html)).

3. **Construction (Polymer / Molecule)**

   * Whether the active layer material is a single small molecule or a polymer.
   * This affects how molecules are processed experimentally, and may correlate with properties like stability or efficiency.

4. **Architecture**

   * The type of device structure used in the solar cell (e.g., bulk heterojunction).
   * Relevant for interpreting performance metrics, but not directly a molecular property.

5. **Complement**

   * The electron acceptor or donor material paired with the molecule in the solar cell.
   * For example, PCBM derivatives are often listed here.
   * This is important for device-level performance

6. **HOMO (Highest Occupied Molecular Orbital)**

   * The energy level of the highest filled electron orbital (in eV).
   * A molecular property that influences how easily the molecule donates electrons.
   * **Prediction target** for your ML task.

7. **LUMO (Lowest Unoccupied Molecular Orbital)**

   * The energy level of the lowest empty electron orbital (in eV).
   * Indicates how easily the molecule can accept electrons.
   * **Prediction target** for your ML task.

8. **Electrochemical gap**

   * The difference between LUMO and HOMO (LUMO − HOMO); in this file it is computed from the two columns above.
   * In the original HOPV data, HOMO and LUMO were measured experimentally, often with cyclic voltammetry. In this course they are synthetic (see the note above).
   * A rough estimate of the energy needed to excite the molecule. (The open-circuit voltage of a device depends mainly on the offset between the HOMO of the donor and the LUMO of the acceptor, see VOC below.)

9. **Optical gap**

   * The band gap inferred from absorption spectra (UV-Vis).
   * Related but not identical to the HOMO-LUMO gap because of excitonic effects.

10. **PCE (Power Conversion Efficiency)**

    * Percentage of sunlight converted into electricity by the device.
    * A device-level metric, depends not only on the molecule but also the architecture and complement.

11. **VOC (Open-Circuit Voltage)**

    * The maximum voltage a device can deliver when no current flows.
    * Influenced by HOMO of donor and LUMO of acceptor.

12. **JSC (Short-Circuit Current Density)**

    * Current per unit area when the circuit is shorted.
    * Related to how well the molecule absorbs light and generates charge carriers.

13. **Fill Factor (FF)**

    * A measure of how “square” the current–voltage curve is.
    * Affects the efficiency of the solar cell.

The rest of the file: after the experimental line, each molecule has one or more conformers (3D structures of the molecule). For each conformer there is the number of atoms, one line per atom with its coordinates, and four lines with properties computed with density functional theory (DFT), one for each computational method (B3LYP, BP86, M06-2X and PBE0). Each of these lines gives the computed HOMO, LUMO and gap in Hartree (1 Hartree = 27.211 eV), followed by the PCE, VOC and JSC that the Scharber model predicts from them. The computed values are the published ones and are not adjusted. They are not targets, and they are not good features either: they are strongly related to the targets (the B3LYP LUMO and the `lumo` column have a correlation of 0.81), but the file gives them only for the HOPV molecules. A model that uses them looks very good on HOPV and cannot predict the lab or test molecules. You do not need them in this course.
